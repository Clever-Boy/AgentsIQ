## 1. Initialize the Intelligent Router


```python
# Initialize the router with intelligent model selection
router = ModelRouter()

print("🧠 Available Models:")
for model, profile in router.profiles.items():
    print(f"  {model}: Cost=${profile['cost']:.3f}, Latency={profile['latency']:.2f}s, Quality={profile['quality']:.2f}")

print(f"\n⚙️ Routing Strategy: {router.strategy}")
print(f"⚖️ Weights: Cost={router.weights['cost']:.1%}, Latency={router.weights['latency']:.1%}, Quality={router.weights['quality']:.1%}")
```

## 2. Create Specialized Search Agents


```python
# Create specialized search agents for different tasks
web_searcher = Agent(
    name="WebSearcher",
    description="Searches the web for information and provides comprehensive answers",
    preferred_model="openai:gpt-4o-mini",
    tools=["web_search", "summarize"]
)

research_agent = Agent(
    name="Researcher",
    description="Conducts deep research and analysis on complex topics",
    preferred_model="anthropic:claude-3-haiku",
    tools=["research", "analyze", "synthesize"]
)

news_agent = Agent(
    name="NewsAgent",
    description="Searches for and analyzes news and current events",
    preferred_model="google:gemini-pro",
    tools=["news_search", "fact_check", "trend_analysis"]
)

print("🤖 Specialized Search Agents Created:")
agents = [web_searcher, research_agent, news_agent]
for agent in agents:
    print(f"  {agent.name}: {agent.description}")
    print(f"    Tools: {agent.tools}")
    print(f"    Preferred Model: {agent.preferred}")
    print()
```
