## 4. Advanced Dashboard & Analytics


```python
# Advanced analytics and dashboard
from agentsiq.decision_store import latest_decisions
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Get comprehensive decision data
all_decisions = latest_decisions(50)
print("📊 Advanced Analytics Dashboard")
print("=" * 60)

if all_decisions:
    df = pd.DataFrame(all_decisions)
    
    # Create comprehensive dashboard
    fig = plt.figure(figsize=(20, 15))
    
    # 1. Model Performance Heatmap
    plt.subplot(3, 3, 1)
    model_performance = df.groupby('chosen').agg({
        'est_cost_chosen': 'mean',
        'latency': 'mean', 
        'quality': 'mean'
    }).round(3)
    
    sns.heatmap(model_performance.T, annot=True, cmap='RdYlBu_r', fmt='.3f')
    plt.title('Model Performance Heatmap')
    plt.xlabel('Models')
    plt.ylabel('Metrics')
    
    # 2. Cost Efficiency Analysis
    plt.subplot(3, 3, 2)
    efficiency = df.groupby('chosen').apply(
        lambda x: x['quality'].mean() / x['est_cost_chosen'].mean()
    ).sort_values(ascending=False)
    
    efficiency.plot(kind='bar', color='lightgreen')
    plt.title('Cost Efficiency (Quality/Cost)')
    plt.xlabel('Models')
    plt.ylabel('Efficiency Ratio')
    plt.xticks(rotation=45)
    
    # 3. Latency Distribution by Model
    plt.subplot(3, 3, 3)
    df.boxplot(column='latency', by='chosen', ax=plt.gca())
    plt.title('Latency Distribution by Model')
    plt.xlabel('Models')
    plt.ylabel('Latency (seconds)')
    plt.xticks(rotation=45)
    
    # 4. Quality vs Cost Scatter
    plt.subplot(3, 3, 4)
    for model in df['chosen'].unique():
        model_data = df[df['chosen'] == model]
        plt.scatter(model_data['est_cost_chosen'], model_data['quality'], 
                   label=model, alpha=0.7, s=60)
    plt.xlabel('Cost ($)')
    plt.ylabel('Quality Score')
    plt.title('Quality vs Cost Trade-off')
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    
    # 5. Usage Trends Over Time
    plt.subplot(3, 3, 5)
    df['timestamp'] = pd.to_datetime(df.get('timestamp', pd.Timestamp.now()))
    usage_trends = df.groupby([df['timestamp'].dt.hour, 'chosen']).size().unstack(fill_value=0)
    usage_trends.plot(kind='line', ax=plt.gca())
    plt.title('Usage Trends by Hour')
    plt.xlabel('Hour of Day')
    plt.ylabel('Usage Count')
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    
    # 6. Strategy Effectiveness
    plt.subplot(3, 3, 6)
    strategy_performance = df.groupby('strategy').agg({
        'est_cost_chosen': 'mean',
        'latency': 'mean',
        'quality': 'mean'
    })
    
    strategy_performance.plot(kind='bar', ax=plt.gca())
    plt.title('Strategy Performance Comparison')
    plt.xlabel('Strategy')
    plt.ylabel('Average Value')
    plt.xticks(rotation=45)
    plt.legend(['Cost', 'Latency', 'Quality'])
    
    # 7. Cost Savings Analysis
    plt.subplot(3, 3, 7)
    # Calculate potential savings vs most expensive model
    max_cost_model = df['est_cost_chosen'].max()
    savings = df.groupby('chosen').apply(
        lambda x: (max_cost_model - x['est_cost_chosen'].mean()) * len(x)
    ).sort_values(ascending=False)
    
    savings.plot(kind='bar', color='orange')
    plt.title('Cost Savings vs Most Expensive Model')
    plt.xlabel('Models')
    plt.ylabel('Total Savings ($)')
    plt.xticks(rotation=45)
    
    # 8. Performance Summary Table
    plt.subplot(3, 3, 8)
    plt.axis('off')
    
    summary_stats = df.groupby('chosen').agg({
        'est_cost_chosen': ['count', 'mean', 'sum'],
        'latency': 'mean',
        'quality': 'mean'
    }).round(4)
    
    summary_text = "📈 Performance Summary:\n\n"
    for model in summary_stats.index:
        count = summary_stats.loc[model, ('est_cost_chosen', 'count')]
        avg_cost = summary_stats.loc[model, ('est_cost_chosen', 'mean')]
        total_cost = summary_stats.loc[model, ('est_cost_chosen', 'sum')]
        avg_latency = summary_stats.loc[model, ('latency', 'mean')]
        avg_quality = summary_stats.loc[model, ('quality', 'mean')]
        
        summary_text += f"{model}:\n"
        summary_text += f"  Uses: {count}, Avg Cost: ${avg_cost:.4f}\n"
        summary_text += f"  Total Cost: ${total_cost:.4f}, Latency: {avg_latency:.2f}s\n"
        summary_text += f"  Quality: {avg_quality:.3f}\n\n"
    
    plt.text(0.1, 0.9, summary_text, transform=plt.gca().transAxes, 
             fontsize=10, verticalalignment='top', fontfamily='monospace')
    
    # 9. ROI Analysis
    plt.subplot(3, 3, 9)
    roi_data = df.groupby('chosen').apply(
        lambda x: (x['quality'].mean() * 100) / (x['est_cost_chosen'].mean() * 1000)
    ).sort_values(ascending=False)
    
    roi_data.plot(kind='bar', color='purple')
    plt.title('ROI Analysis (Quality per $1000)')
    plt.xlabel('Models')
    plt.ylabel('ROI Score')
    plt.xticks(rotation=45)
    
    plt.tight_layout()
    plt.show()
    
    # Print detailed analytics
    print(f"\n📊 Detailed Analytics:")
    print(f"   Total Decisions Analyzed: {len(df)}")
    print(f"   Unique Models Used: {df['chosen'].nunique()}")
    print(f"   Total Cost: ${df['est_cost_chosen'].sum():.4f}")
    print(f"   Average Latency: {df['latency'].mean():.2f}s")
    print(f"   Average Quality: {df['quality'].mean():.3f}")
    print(f"   Most Used Model: {df['chosen'].mode().iloc[0]}")
    print(f"   Most Cost-Effective: {efficiency.idxmax()}")
    print(f"   Highest ROI: {roi_data.idxmax()}")
    
else:
    print("⚠️ No decision data available. Run some collaborative tasks first to see the advanced dashboard.")
```

```python
# Create specialized search agents for different tasks
web_searcher = Agent(
    name="WebSearcher",
    description="Searches the web for information and provides comprehensive answers",
    preferred_model="openai:gpt-4o-mini",
    tools=["web_search", "summarize"]
)

research_agent = Agent(
    name="Researcher",
    description="Conducts deep research and analysis on complex topics",
    preferred_model="anthropic:claude-3-haiku",
    tools=["research", "analyze", "synthesize"]
)

news_agent = Agent(
    name="NewsAgent",
    description="Searches for and analyzes news and current events",
    preferred_model="google:gemini-pro",
    tools=["news_search", "fact_check", "trend_analysis"]
)

print("🤖 Specialized Search Agents Created:")
agents = [web_searcher, research_agent, news_agent]
for agent in agents:
    print(f"  {agent.name}: {agent.description}")
    print(f"    Tools: {agent.tools}")
    print(f"    Preferred Model: {agent.preferred}")
    print()
```
