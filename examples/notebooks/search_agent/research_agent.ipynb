## 1. Initialize Research System


```python
# Initialize the router and create research agents
router = ModelRouter()

# Create specialized research agents
primary_researcher = Agent(
    name="PrimaryResearcher",
    description="Conducts comprehensive research on complex topics",
    preferred_model="anthropic:claude-3-haiku",
    tools=["research", "analyze", "synthesize"]
)

fact_checker = Agent(
    name="FactChecker",
    description="Verifies facts and checks information accuracy",
    preferred_model="openai:gpt-4o-mini",
    tools=["fact_check", "verify", "cross_reference"]
)

synthesizer = Agent(
    name="Synthesizer",
    description="Synthesizes information into comprehensive reports",
    preferred_model="google:gemini-pro",
    tools=["synthesize", "summarize", "structure"]
)

print("🔬 Research System Initialized:")
print(f"  Primary Researcher: {primary_researcher.name}")
print(f"  Fact Checker: {fact_checker.name}")
print(f"  Synthesizer: {synthesizer.name}")
print(f"  Router Strategy: {router.strategy}")
```
